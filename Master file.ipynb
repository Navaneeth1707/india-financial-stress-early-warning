```python
# Combine both CSV files into one master file
import pandas as pd

# Read both files
financial = pd.read_csv(
    r"D:\workfiles\projects resume 1234567890\india_financial_stress\04_tableau\financial_stress_tableau.csv"
)

ml = pd.read_csv(
    r"D:\workfiles\projects resume 1234567890\india_financial_stress\04_tableau\ml_predictions_tableau.csv"
)

# Merge on ticker and year
combined = financial.merge(
    ml[['ticker','pred_year','predicted_risk',
        'prob_safe','prob_grey','prob_distress',
        'confidence_pct','is_correct']],
    left_on=['ticker','report_year'],
    right_on=['ticker','pred_year'],
    how='left'
)

# Save as one master file
combined.to_csv(
    r"D:\workfiles\projects resume 1234567890\india_financial_stress\04_tableau\master_tableau.csv",
    index=False
)

print(f"Master file created!")
print(f"Rows: {len(combined)}")
print(f"Columns: {combined.columns.tolist()}")
```
